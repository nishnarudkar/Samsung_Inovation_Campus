```python
''' You are given a set of 2D points corresponding the vertices of a geometric shape
1. Create a numpy array P of shape(N,2) contaning atleast 6 dimensional points of your choice 
2.Define
        A rotation matrix for a rotation of 45degree
        A scaling matrix that scales the x-axis by 2 and the y-axis by 0.5
3. Combine the rotation and scaling into a single linear transformation matrix
4. Apply the combined transformation to all points using numpy matrix operations(no loops).
5. Compute and print:
            The transformed coordinates
            The Eucledian distance between each original point and its transformed point'''

import numpy as np
P = np.array([
    [1, 2],
    [3, 5],
    [6, 4],
    [8, 7],
    [5, 9],
    [2, 6]
])

print("Original Points (P):\n", P)

theta = 45  # degrees
theta_rad = np.deg2rad(theta)

R = np.array([
    [np.cos(theta_rad), -np.sin(theta_rad)],
    [np.sin(theta_rad),  np.cos(theta_rad)]
])

S = np.array([
    [2, 0],
    [0, 0.5]
])

T = R @ S

print("\nCombined Transformation Matrix:\n", T)

P_transformed = P @ T.T  

print("\nTransformed Coordinates:\n", P_transformed)

# Compute Euclidean distance between original and transformed points
distances = np.linalg.norm(P_transformed - P, axis=1)

print("\nEuclidean Distance for each point:\n", distances)
```

```python
''' 2.Consider the following real symmetric matrix A:
A=[[4,1,1],
    [1,3,0],
    [1,0,2]]
Tasks:
1. Represent the matrix A using Numpy
2. Compute the eigen values and eigen vectors of A.
3. Verify nummerically that A v = lamda v for one eigenvalue-eigenvector pair.
4. Check whether the eignevectors are orthogonal.
5. Sort the eigenvalues in descending order and rearrange eigenvectors accordingly.
'''
import numpy as np

A = np.array([[4,1,1],
              [1,3,0],
              [1,0,2]])

eigenvalues,eigenvectors = np.linalg.eigh(A)

print(eigenvalues)
print(eigenvectors)

v = eigenvectors[:,0]
lam = eigenvalues[0]

left = A@v
right = lam*v

print(left)
print(right)

orthogonality_check = eigenvectors.T @ eigenvectors
print(orthogonality_check)

idx= np.argsort(eigenvalues)[::-1]
eigenvalues_sorted = eigenvalues[idx]
eigenvectors_sorted = eigenvectors[:,idx]

print(eigenvalues_sorted)
print(eigenvectors_sorted)
```
