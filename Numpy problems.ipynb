```python
# Write a program using numpy to create a zero vector of size n(user defined) and display it

import numpy as np
n = int(input("Enter the size of the zero vector: "))
zero_vector = np.zeros(n)
print("Zero Vector of size", n, ":", zero_vector)
```

```python
# Create a one vector of size 7 using numpy. Print the vector and shape

import numpy as np
one_vector = np.ones(7)
print(one_vector)
print(one_vector.shape)
```

```python
#Create a one vector of length 10,reshape it into a 2 by 5 matrix and compute the row-wise and column-wise sums

import numpy as np

one_vector = np.ones(10)
matrix = one_vector.reshape(2,5)

row_sums = np.sum(matrix,axis=1)
print(row_sums)
column_sums = np.sum(matrix,axis=0)
print(column_sums)
```

```python
# Generate a square matrix of size n by n with random integers

import numpy as np
n = int(input("Enter a number: "))
random_matrix = np.random.randint(0,10,size=(n,n))
print(random_matrix)
```

```python
""" The prices of three grocery items rice, milk and bread are 120rs,45rs and 30rs per item respectively. A customer wants to buy 3 packets of rice,4 bottles of milk and 5 loaves of bread 
1. Represent the item prices and quantities purchased as numpy vectors p and q respectively
2. Using appropriate numpy vector operations, compute the total cost of the purchase
"""
import numpy as np
p = np.array([120,45,30])
q = np.array([3,4,5])

total_cost = np.dot(p,q)
print("Prices vector: ",p)
print("Quantity vector: ",q)
print("Total cost: ",total_cost)
```
