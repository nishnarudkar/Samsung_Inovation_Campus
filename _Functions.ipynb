```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 400)

y1 = 3 ** (x - 1) + 2
y2 = 2 ** (-x - 2)

plt.figure()
plt.plot(x, y1, label="y = 3^(x-1) + 2")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of y = 3^(x-1) + 2")
plt.legend()
plt.grid(True)
plt.show()

plt.figure()
plt.plot(x, y2, label="y = 2^(-x - 2)")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of y = 2^(-x - 2)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-0.99, 5, 400)   # x > -1

y = np.log2(x + 1) - 1

plt.figure()
plt.plot(x, y, label="y = log₂(x+1) - 1")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of y = log₂(x+1) - 1")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x= np.linspace(-5,5,400)
y=1/(1+np.exp(-x))

plt.figure()
plt.plot(x,y,label="Sigmoid Function")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Sigmoid Function")
plt.legend()
plt.grid(True)
plt.show()
```
