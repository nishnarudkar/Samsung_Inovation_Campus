```python
# Find the derivative of the constant function: f(x) = x**2020 + 3*x**3 -1

import sympy as sp

x= sp.Symbol('x')
fx=x**2020 + 3*x**3 -1
print("f(x) = ",fx)
print("f'(x)= ",sp.diff(fx,x))
```

```python
# Find the derivative of the constant function: 

x = sp.Symbol('x')
gx= x/(x+1)
print("g'(x)= ", sp.simplify(sp.diff(gx,x)))
```

```python
# Find the derivative: f(x) = exp(x)/x ( but x!= 0)

x = sp.Symbol('x')
fx= sp.exp(x)/x
print("f'(x) = ",sp.diff(fx,x))
```

```python
# (b) g(x) = 3log2(x) - xlogx

x = sp.Symbol('x')
gx = 3*sp.log(x,2) - x*sp.log(x)
print("g'(x) = ", sp.simplify(sp.diff(gx,x)))
```

```python
# f(x) = (x**2 + 2)**3
x = sp.Symbol('x')
fx = (x**2+2)**3
print("f'(x) = ", sp.diff(fx,x))
```

```python
# g(x) = 1/(x**4+1)**2

x= sp.Symbol('x')
gx = 1/(x**4 + 1)**2
print("g'(x) = ", sp.diff(gx,x))
```
