```python
''' 1.Find the first/second partial derivatives fx,fy,fxx,fyy,fxy,fyx for the following function using SymPy
f(x,y) = exp(x**3 + 2*y**3)
'''

import sympy as sp
x,y = sp.symbols('x y')
f = sp.exp(x**3 + 2*y**3)
print("f(x,y) = ", f)
# Calcualte fx
fx = sp.diff(f,x)
print("fx = ",fx)
# Calculate fy
fy = sp.diff(f,y)
print("fy = ",fy)
# Calulate fxx
fxx = sp.diff(fx,x)
print("fxx = ",fxx)
# Calulate fyy
fyy = sp.diff(fy,y)
print("fyy = ", fyy)
# Calulate fxy
fxy = sp.diff(fx,y)
print("fxy = ", fxy)
# Calulate fyx
fyx = sp.diff(fy,x)
print("fyx = ", fyx)

```

```python
''' 2. Find the derivative that differentiates the following function using Sympy. In this expression  k, a , and b are constants not variables.
1. f(x) = x**5 - 1 
'''

import sympy as sp
x= sp.Symbol('x')
fx = x**5 - 1
print("f(x) = ",sp.diff(fx,x))
```

```python
# 2. f(x)= log(x**3-7k)

x,k = sp.symbols('x k')
f = sp.log(x**3- 7*k)
print("f(x) = ", sp.diff(f,x))
```

```python
# 3.f(x) = exp(ax**b)

x,a,b = sp.symbols(' x a b')
f = sp.exp(a*x**b)
print("f'(x) = ",sp.simplify( sp.diff(f,x)))
```

```python
'''3. Find the eigenvalues of the following matrix using the characteristic equation.
D=[[2,1],[1,2]]
'''

import sympy as sp
lam = sp.Symbol('lam')
D = sp.Matrix([[2, 1],
               [1, 2]])
char_eq = sp.det(D - lam*sp.eye(2))
print("Characteristic Equation:", char_eq)
eigenvalues = sp.solve(char_eq, lam)
print("Eigenvalues:", eigenvalues)
```

```python
''' 4. The log5 value is about 1.61 and the log4 value is about 1.39. At this time, find the value of log20'''

import sympy as sp
log5 = sp.Float(1.61)
log4 = sp.Float(1.39)
log20 = log5 + log4

print("log(20) ≈", log20)
```
