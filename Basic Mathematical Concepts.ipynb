```python
n=10
a = 5
d=3

sum1=0
for n in range (1,n+1):
    sum1= sum1 + (a+(n-1)*d)
print("S10 = ",sum1, "by summation")
```

```python
#Sum 
n = 10
x = 7
total=0

for _ in range(n):  
    total += x 

print(total)
```

```python
#Product
n=10
x=7
product=1

for _ in range(n):
    product*=x
print(product)
```

```python
n=10
sum1=0

for k in range(1,n+1):
    sum1=sum1+(k**2-k)
print("Sum = ",sum1, "by summation")
```
